# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
````

# 2- What is Word2vec?

Word2Vec consists of models for generating word embedding. These models are two-layer neural networks having one input layer, one hidden layer, and one output layer.

Word2Vec utilizes two architectures :
1. CBOW (Continuous Bag of Words)
2. **Skip Gram**

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = ['This is the first document.',
        'This document is the second document.',
        'And this is the third one.',
        'Is this the first document?']

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(docs)

sim = cosine_similarity(vectors)
pd.DataFrame(sim, index=docs, columns=docs).round(3)
````

- Sentence 1 and 4 have similarity **1.0** because after tokenizing and lower casing they have exactly the same words (this, is, the, first, document), only the order and punctuation are different. TF-IDF doesn't care about word order.
- Sentence 2 is close to 1 and 4 because they share "document", "this", "is", "the".
- Sentence 3 is the least similar to the others since it only shares common words like "this", "is", "the".

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
docs2 = ['data science is one of the most important fields of science',
         'this is one of the best data science courses',
         'data scientists analyze data']

tfidf2 = TfidfVectorizer()
result2 = tfidf2.fit_transform(docs2)

df_tfidf = pd.DataFrame(result2.toarray(), columns=tfidf2.get_feature_names_out(),
                        index=['doc1', 'doc2', 'doc3'])
df_tfidf.round(3)
````

````python
# top 3 important words in each sentence
for i, row in df_tfidf.iterrows():
    print(i, ':', row.sort_values(ascending=False).head(3).round(3).to_dict())
````

````python
# idf of each word
pd.Series(tfidf2.idf_, index=tfidf2.get_feature_names_out()).sort_values().round(3)
````

- "data" appears in all 3 sentences so it has the lowest IDF (1.0). It's not useful for telling the sentences apart, but in doc 3 it still gets the highest score (0.641) because it's repeated twice in a very short sentence.
- Words that appear in only one sentence have the highest IDF (1.693), like "fields" and "important" in doc 1, "best" and "courses" in doc 2, and "scientists" and "analyze" in doc 3. These are the important words for each sentence.
- In doc 1 "science" and "of" are on top because they appear twice. "of" is a stop word, so if we removed stop words first the result would make more sense.

````python
import re
import gensim
from nltk.tokenize import word_tokenize

simpsons = pd.read_csv('dataset/simpsons_script_lines.csv', low_memory=False)
print(simpsons.shape)
simpsons.head()
````

````python
# some lines don't have spoken words (only actions), so drop them
print(simpsons['spoken_words'].isna().sum())
simpsons = simpsons.dropna(subset=['spoken_words'])
print(simpsons.shape)
````

````python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons['clean'] = simpsons['spoken_words'].astype(str).apply(clean_text)
simpsons[['spoken_words', 'clean']].head()
````

````python
tokens = [word_tokenize(line) for line in simpsons['clean']]
print(len(tokens))
print(tokens[:3])
````

````python
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count = 1, vector_size = 100, window = 5, sg = 1)

print("Vocabulary size:", len(Skip_gram_model.wv))
````
